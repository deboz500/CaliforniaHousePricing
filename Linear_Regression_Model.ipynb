```python
#identifying independent and dependent features
# We know price is the only dependent feature here
X=df.iloc[:,:-1]
Y=df.iloc[:,-1]
```

```python
X.head()
```

```python
Y
```

```python
#Train Test Split
from sklearn.model_selection import train_test_split
X_train,X_test,Y_train,Y_test=train_test_split(X,Y,test_size=0.2,random_state=42)
```

```python
X_train
```

```python
X_test
```

```python
#Standardizing the dataset
from sklearn.preprocessing import StandardScaler
scaler=StandardScaler()
```

```python
X_train=scaler.fit_transform(X_train)
X_test=scaler.transform(X_test)
```

```python
X_train
```

Model Training

```python
from sklearn.linear_model import LinearRegression
```

```python
regression=LinearRegression()
```

```python
regression.fit(X_train,Y_train)
```

```python
#printing the coefficients and the intercept
print(regression.coef_)#this will be equal to no. of independent features
```

```python
print(regression.intercept_)
```

```python
#parameters for model training
regression.get_params()
```

```python
#Prediction with Test Data
reg_pred=regression.predict(X_test)
```

```python
reg_pred
```

```python
#scatterplot for prediction
plt.scatter(Y_test,reg_pred)
```

The above scatterplot shows a linear relationship. This indicates the model has performed well.

```python
##Calculating residuals(error in prediction)
residuals=Y_test-reg_pred
```

```python
residuals
```

```python
#Prediction with Residuals
sns.displot(residuals,kind='kde')
```

```python
#Scatterplot for prediction and residuals
plt.scatter(reg_pred,residuals)
```

The above scatterplot shows the following behaviour because the house prices in California are capped to 5.0

Performance Metrics

```python
from sklearn.metrics import mean_squared_error
from sklearn.metrics import mean_absolute_error
```

```python
print(mean_absolute_error(Y_test,reg_pred))
print(mean_squared_error(Y_test,reg_pred))
print(np.sqrt(mean_squared_error(Y_test,reg_pred)))#rms error
```

```python
#R square and adjusted R square
from sklearn.metrics import r2_score
score=r2_score(Y_test,reg_pred)
print(score)
1-(1-score)*(len(Y_test)-1)/(len(Y_test)-X_test.shape[1]-1)

#these values closer to 1 means predicition model working well
```
